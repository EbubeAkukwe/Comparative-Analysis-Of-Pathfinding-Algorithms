# Analysis of Experiment 1 Sets

The analysis from this notebook will make up the data shown in Table 4.1.2 and some of the data in Table 4.3.3  (see Paper)

```python
import pandas as pd
import numpy as np

TOTAL_RUNS = 30
```

#### Analysis of Experiment 1 Set 1
Computation of Mean ± Standard deviation of density, execution time in milliseconds (ms),nodes expanded,length of path, and memory in bytes (B)

```python
df = df = pd.read_csv("../dataset/sets/experiment_1/exp2_set1.csv")

df = df.replace('NULL', np.nan) 

group_cols = ['Algorithm', 'Grid_Size', 'Density']
metric_cols = ['Execution_Time', 'Nodes_Expanded', 'Path_Length', 'Memory_B']

for col in metric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df['Is_Success'] = df[metric_cols].notnull().all(axis=1).astype(int)


stats = df.groupby(group_cols).agg(
    null_count=('Is_Success', lambda x: (x == 0).sum())
)

stats['Success_Rate_%'] = 100 - ((stats['null_count'] / TOTAL_RUNS) * 100)

df_success = df[df['Is_Success'] == 1].copy()
summary = df_success.groupby(group_cols)[metric_cols].agg(['mean', 'std'])

formatted_summary = pd.DataFrame(index=summary.index)
for col in metric_cols:
    formatted_summary[col] = (
        summary[col]['mean'].map('{:.3f}'.format) + 
        " ± " + 
        summary[col]['std'].map('{:.3f}'.format)
    )

formatted_summary['Success_Rate_%'] = stats['Success_Rate_%'].map('{:.1f}%'.format)

# Display final table
display(formatted_summary)
```
