### Statistical Tests for Experiment 2

```python
import scipy.stats as stats

import pandas as pd
import numpy as np
```

```python
#Prepare data

def prepare_data(csv_path):
    df = pd.read_csv(csv_path)
    df = df.replace([float('inf'), float('-inf')], float('nan')).dropna()
    # Extract only the numeric value from "Mean±Std" columns
    # Example: '0.549±0.477' -> 0.549
    for col in df.columns:
        if df[col].dtype == 'object' and '±' in str(df[col].iloc[0]):
            df[col] = df[col].apply(lambda x: float(str(x).split('±')[0]))
    return df
```

```python
# Data Preprocessing

df = prepare_data('../dataset/sets/exp2_sets.csv')

# 2. Re-check for any remaining NaNs or Infs explicitly
df = df.dropna(subset=['Execution_Time', 'Path_Length', 'Optimality_Ratio'])
df = df.replace([np.inf, -np.inf], np.nan).dropna(subset=['Execution_Time'])

display(df)
```

```python
# One-Way ANOVA
groups = [group['Execution_Time'].values for name, group in df.groupby('Heuristic')]
f_stat, p_val = stats.f_oneway(*groups)
print(f"One-Way ANOVA: F={f_stat:.4f}, p={p_val:.4f}")
```

```python
# Eta Squared (η²) Calculation
ss_total = np.sum((df['Execution_Time'] - df['Execution_Time'].mean())**2)
ss_between = np.sum([len(g) * (g['Execution_Time'].mean() - df['Execution_Time'].mean())**2 for n, g in df.groupby('Heuristic')])
eta_sq = ss_between / ss_total
print(f"Execution Time Effect Size (η²): {eta_sq:.4f}")
```

```python
# Eta Squared (η²) Calculation
ss_total = np.sum((df['Nodes_Expanded'] - df['Nodes_Expanded'].mean())**2)
ss_between = np.sum([len(g) * (g['Nodes_Expanded'].mean() - df['Nodes_Expanded'].mean())**2 for n, g in df.groupby('Heuristic')])
eta_sq = ss_between / ss_total
print(f"Nodes Expanded Effect Size (η²): {eta_sq:.4f}")
```

```python
# Eta Squared (η²) Calculation
ss_total = np.sum((df['Optimality_Ratio'] - df['Optimality_Ratio'].mean())**2)
ss_between = np.sum([len(g) * (g['Optimality_Ratio'].mean() - df['Optimality_Ratio'].mean())**2 for n, g in df.groupby('Heuristic')])
eta_sq = ss_between / ss_total
print(f"Optimality Ratio Effect Size (η²): {eta_sq:.4f}")
```

Statistical Test Summary

One-Way ANOVA: F=106.0409, p=0.0000

Execution Time Effect Size (η²): 0.5451 (strong)

Nodes Expanded Effect Size (η²): 0.5872 (strong)

Optimality Ratio Effect Size (η²): 0.0027 (very weak, we can neglect)
