### Statistical Tests for Experiment 3

```python
import scipy.stats as stats

import pandas as pd
import numpy as np
```

```python
# Data Preprocessing

df = pd.read_csv('../dataset/sets/exp3_sets.csv')
df['Optimality_Ratio'] = df['Path_Length'].apply(lambda x: float(x/49))
df['%_Path_Found'] = df['Path_Length'].apply(lambda x: float((49/x)*100))
df['%_Grid_Explored'] = df['Nodes_Expanded'].apply(lambda x: float((x/2500)*100))
df = df.replace([float('inf'), float('-inf')], float('nan')).dropna()


display(df)
```

Statistical Tests

```python
# Define groups
g1 = df[df['Movement_Model'] == '4-dir']['Execution_Time']
g2 = df[df['Movement_Model'] == '8-dir']['Execution_Time']

# Test for Normality (Shapiro-Wilk)
if stats.shapiro(g1).pvalue > 0.05 and stats.shapiro(g2).pvalue > 0.05:
    stat, p = stats.ttest_ind(g1, g2)
    print(f"Independent T-Test: stat={stat:.4f}, p={p:.4f}")
else:
    stat, p = stats.mannwhitneyu(g1, g2)
    print(f"Mann-Whitney U Test for Execution Time: stat={stat:.4f}, p={p:.4f}")
```

```python
# Define groups
g1 = df[df['Movement_Model'] == '4-dir']['Nodes_Expanded']
g2 = df[df['Movement_Model'] == '8-dir']['Nodes_Expanded']

# Test for Normality (Shapiro-Wilk)
if stats.shapiro(g1).pvalue > 0.05 and stats.shapiro(g2).pvalue > 0.05:
    stat, p = stats.ttest_ind(g1, g2)
    print(f"Independent T-Test: stat={stat:.4f}, p={p:.4f}")
else:
    stat, p = stats.mannwhitneyu(g1, g2)
    print(f"Mann-Whitney U Test for Node Expansion: stat={stat:.4f}, p={p:.4f}")
```
