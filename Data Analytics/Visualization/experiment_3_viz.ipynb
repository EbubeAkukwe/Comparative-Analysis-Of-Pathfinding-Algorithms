### Experiment 3 Visualizations

The Visualizations in this nnotebook will make up the figures shown in Experiment 3 results

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_pathCost = pd.read_csv('../Experiment/dataset/experiment3_pathCost.csv')

display(df_pathCost)
```

Effect of Movement model on Path Optimality

```python
# Create the Horizontal Dot Plot
plt.figure(figsize=(10, 6))

# Use stripplot (which is essentially a dot plot)
sns.stripplot(data=df_pathCost, x='Optimality_Ratio', y='Algorithm', hue='Movement', 
              dodge=True, size=10, marker='o', jitter=False)

# Add a reference line at 1.0 (Optimal)
plt.axvline(1.0, color='red', linestyle='--', label='Optimal Path (1.0)')

# Add grid lines for readability
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.title('Figure: Summary of Optimality Ratio by Algorithm and Movement')
plt.xlabel('Optimality Ratio (Lower is Better)')
plt.ylabel('Algorithm')
plt.legend(title='Movement Model', loc='upper right')
plt.tight_layout()

plt.show()
```

Effect of Movement model on Node expansion

```python
df_nodeExpanded = pd.read_csv('../Experiment/dataset/experiment3_nodeExpanded.csv')

display(df_nodeExpanded)
```

```python
# Get the mean node expanded
df_nodeExpanded['Mean_Nodes'] = df_nodeExpanded['Nodes_Expanded'].apply(lambda x: float(x.split('±')[0]))

# Create the plot
plt.figure(figsize=(10, 6))

# Grouped Bar Chart to show the effectiveness of algorithms
sns.barplot(data=df_nodeExpanded, x='Algorithm', y='Mean_Nodes', hue='Movement',)

plt.title('Nodes Expanded by Algorithm and Movement Model (20% Density)')
plt.ylabel('Mean Nodes Expanded')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```
